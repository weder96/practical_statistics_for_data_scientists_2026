# Complete Guide to the SciPy Library

**SciPy** (*Scientific Python*) is the core open-source library for scientific, mathematical, and engineering computing in Python. Built directly on top of **NumPy** arrays, it provides high-performance implementations for complex numerical algorithms — many of them written internally in C, C++, and Fortran (wrapping classic libraries such as BLAS, LAPACK, MINPACK, and ODEPACK).

---

## 1. Architecture and Installation

Unlike libraries where a single `import` exposes everything, SciPy is organized into **independent submodules**. For performance and memory management reasons, submodules should be imported explicitly.

```python
# Uncomment the line below if you need to install SciPy in the current environment
# !pip install scipy

# Recommended way to import
import numpy as np
from scipy import stats, optimize, integrate, linalg
import scipy

print(f'Installed SciPy version: {scipy.__version__}')
```

---

## 2. Complete Map of Submodules

| Submodule | Main Purpose | Example Features |
| :--- | :--- | :--- |
| **`scipy.stats`** | Statistics and probability | 100+ continuous/discrete distributions, hypothesis tests (*t-test*, ANOVA, Chi-square, Shapiro-Wilk, Mann-Whitney), correlations. |
| **`scipy.optimize`** | Optimization and root finding | Unconstrained and constrained minimization (`minimize`), curve fitting (`curve_fit`), linear programming (`linprog`), root finding (`root`). |
| **`scipy.integrate`** | Numerical integration and ODEs | Single and multiple definite integrals (`quad`, `dblquad`), solving Ordinary Differential Equations (`solve_ivp`). |
| **`scipy.linalg`** | Advanced linear algebra | Decompositions (LU, QR, Cholesky, SVD, Schur), eigenvalues/eigenvectors, solving linear systems, matrix exponentials. |
| **`scipy.interpolate`** | Data interpolation | Cubic splines (`CubicSpline`), 1D and N-dimensional linear/polynomial interpolation (`interp1d`, `griddata`). |
| **`scipy.signal`** | Signal processing | Digital filters (Butterworth, Chebyshev), convolution, spectrograms, peak detection (`find_peaks`). |
| **`scipy.fft`** | Fourier transforms | 1D/2D/N-D FFT, discrete cosine transform (DCT) and sine transform (DST). |
| **`scipy.sparse`** | Sparse matrices and graphs | CSR, CSC, COO formats for large matrices with many zeros; graph algorithms in `scipy.sparse.csgraph` (Dijkstra, Floyd-Warshall). |
| **`scipy.spatial`** | Computational geometry | Distance calculations (`cdist`, `pdist`), `KDTree`, Voronoi diagrams, Delaunay triangulation, convex hull (`ConvexHull`). |
| **`scipy.ndimage`** | N-D image processing | Gaussian and morphological filters, rotation, resizing, and segmentation of multidimensional arrays. |
| **`scipy.cluster`** | Clustering | Vector quantization / K-Means (`vq`) and hierarchical clustering with dendrograms (`hierarchy`). |
| **`scipy.special`** | Special mathematical functions | Bessel, Gamma, and Beta functions, error function (`erf`), orthogonal polynomials (Legendre, Hermite). |
| **`scipy.io`** | File input and output | Reading/writing MATLAB (`.mat`), audio (`.wav`), Fortran, Matrix Market, and ARFF files. |
| **`scipy.constants`** | Physical and mathematical constants | Speed of light ($c$), Planck's constant ($h$), Avogadro's number, unit conversions. |

## 3. Practical Examples of the Most Used Modules

### 3.1 Statistics and Hypothesis Testing (`scipy.stats`)
Checking whether there is a statistically significant difference between two independent samples using Student's *t*-test:

```python
group_a = np.array([22.1, 23.4, 21.8, 24.0, 22.9])
group_b = np.array([25.2, 26.1, 24.8, 25.9, 26.5])

t_stat, p_value = stats.ttest_ind(group_a, group_b)
print(f't-statistic: {t_stat:.4f} | p-value: {p_value:.4f}')
```

### 3.2 Optimization and Curve Fitting (`scipy.optimize`)
Finding the minimum point of the function $f(x) = (x - 3)^2 + 5$:

```python
func = lambda x: (x - 3)**2 + 5
result = optimize.minimize(func, x0=0)

print(f'Minimum at x = {result.x[0]:.4f} | f(x) = {result.fun:.4f}')
```

### 3.3 Numerical Integration (`scipy.integrate`)
Computing the definite integral $\int_{0}^{\pi} \sin(x)\,dx$:

```python
value, estimated_error = integrate.quad(np.sin, 0, np.pi)
print(f'Result: {value:.6f} (estimated error: {estimated_error:.2e})')
```

### 3.4 Linear Algebra (`scipy.linalg`)
Solving the linear system $Ax = b$:

```python
A = np.array([[3, 2], [1, 4]])
b = np.array([7, 9])

x = linalg.solve(A, b)
print('Solution x:', x)
```

## 4. Where SciPy Fits in the Python Ecosystem

* **NumPy vs. SciPy:** **NumPy** provides the foundational data structure (`ndarray`) and core vectorized operations. **SciPy** builds on NumPy arrays to deliver full-featured scientific algorithms. Even where overlap exists (such as `numpy.linalg` and `scipy.linalg`), the SciPy version is typically more comprehensive and compiled with full BLAS/LAPACK optimizations.
* **Foundation for other libraries:** packages such as **Scikit-learn** (especially sparse matrices and optimizers), **Statsmodels**, and **Pandas** rely directly on SciPy under the hood.
* **When to choose SciPy:** ideal for mathematical modeling, numerical simulations, engineering, signal processing, and rigorous statistical testing. For high-level predictive Machine Learning, prefer **Scikit-learn**; for tabular data manipulation, use **Pandas** or **Polars**.
