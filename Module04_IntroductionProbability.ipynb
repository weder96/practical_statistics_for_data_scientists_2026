## **Conceitos**

### **Experimento aleatório**
Fenômeno cujo resultado não pode ser previsto: se o experimento for repetido nas mesmas condições, o resultado pode ser diferente.

### **Espaço amostral**
O espaço amostral $S$ é o conjunto de todos os resultados possíveis do experimento.

```python
# Espaço amostral com 60 resultados possíveis (0 a 59)
S = list(range(60))
len(S)
```

### **Eventos**
Um evento é um subconjunto do espaço amostral, ou seja, um conjunto de resultados possíveis do experimento.

### **Operações com eventos**

* `União:` o evento $A ∪ B$ reúne todos os resultados que pertencem a $A$, a $B$ ou a ambos.

* `Interseção:` o evento $A ∩ B$ reúne os resultados que pertencem a $A$ e a $B$ ao mesmo tempo.

* `Complementar:` o evento $A^c$ reúne todos os resultados do espaço amostral que não pertencem a $A$.

### **Eventos independentes**
A ocorrência do evento $B$ não altera a probabilidade de ocorrência do evento $A$.

### **Eventos mutuamente excludentes**
Os eventos $A$ e $B$ não possuem nenhum elemento em comum, ou seja, não podem ocorrer simultaneamente ($A ∩ B = ∅$).

## **Definição de probabilidade**
A probabilidade mede a chance de ocorrência de um evento $A$ dentro do espaço amostral. Quando todos os resultados são igualmente prováveis:

$P(A) = \frac{n_{A}}{n}$

sendo $n_A$ o número de resultados favoráveis a $A$ e $n$ o total de resultados possíveis.

```python
# Lançamento de um dado: probabilidade de sair número par (em %)
dados = [1, 2, 3, 4, 5, 6]
A = [2, 4, 6]

probabilidade = len(A)/len(dados) * 100
probabilidade
```

## **Regras básicas**

1.   `Variação da probabilidade:` a probabilidade está sempre entre 0 e 1 (0% a 100%), $0 \leq P(A) \leq 1$
2.   `Probabilidade do espaço amostral:` $P(S) = 1$
3.   `Probabilidade do conjunto vazio:` $P(∅) = 0$
4.   `Regra da adição:` $P(A ∪ B) = P(A) + P(B) - P(A ∩ B)$. Se $A$ e $B$ forem mutuamente excludentes, $P(A ∪ B) = P(A) + P(B)$
5.   `Eventos complementares:` $P(A^c) = 1 - P(A)$
6.   `Regra da multiplicação para eventos independentes:` $P(A ∩ B) = P(A)\cdot P(B)$

## **Probabilidade Condicional**
Quando os eventos não são independentes, usamos a probabilidade condicional: dado que $B$ ocorreu, qual a probabilidade de $A$ ocorrer?

$P(A|B) = \frac{P(A ∩ B)}{P(B)}$

* `Regra da multiplicação:` $P(A ∩ B) = P(A)\cdot P(B|A)$, que pode ser estendida para três eventos: $P(A ∩ B ∩ C) = P(A)\cdot P(B|A)\cdot P(C|A ∩ B)$

## **Teorema de Bayes**
Permite recalcular (atualizar) a probabilidade de um evento quando novas informações ficam disponíveis. Sendo $B_1, ..., B_n$ eventos mutuamente excludentes que formam o espaço amostral:

$P(B_{i} | A) = \frac{P(A|B_i)\cdot P(B_i)}{\sum_{j=1}^{n}P(A|B_j)\cdot P(B_j)}$

## **Análise Combinatória**

Ferramentas para contar o número de resultados possíveis quando o espaço amostral é grande.

* `Arranjo:` número de maneiras de escolher $k$ elementos entre $n$, quando a ordem importa. $A_{n,k} = \frac{n!}{(n-k)!}$
* `Combinação:` número de maneiras de escolher $k$ elementos entre $n$, quando a ordem não importa. $C_{n,k} = \frac{n!}{k!(n-k)!}$
* `Permutação:` número de maneiras de ordenar $n$ elementos. $P_n = n!$

```python
# Exemplo: 10 candidatos para 3 vagas
n, k = 10, 3

print(f"Arranjo (a ordem importa): {math.perm(n, k)}")
print(f"Combinação (a ordem não importa): {math.comb(n, k)}")
print(f"Permutação de 3 elementos: {math.factorial(3)}")
```

Pequena classe para calcular as probabilidades vistas acima a partir de listas de eventos.

```python
class universo:
    def __init__(self, eventoA, espaco_amostral, eventoB = None, eventoC = None):
        self.eventoA = eventoA
        self.eventoB = eventoB
        self.eventoC = eventoC
        self.espaco_amostral = espaco_amostral

    def prob(self):
        return len(self.eventoA) / len(self.espaco_amostral)

    def intercet(self):
        return list(set(self.eventoA).intersection(self.eventoB))

    def union(self):
        return list(set(self.eventoA).union(self.eventoB))

    def adicao_multual(self):
        # P(A ∪ B) = P(A) + P(B) - P(A ∩ B)
        prob_a = universo(self.eventoA, self.espaco_amostral).prob()
        prob_b = universo(self.eventoB, self.espaco_amostral).prob()
        prob_ab = universo(self.intercet(), self.espaco_amostral).prob()
        return prob_a + prob_b - prob_ab

    def adicao_excludantes(self):
        # P(A ∪ B) = P(A) + P(B), válida só se A e B forem mutuamente excludentes
        prob_a = universo(self.eventoA, self.espaco_amostral).prob()
        prob_b = universo(self.eventoB, self.espaco_amostral).prob()
        return prob_a + prob_b

    def complement(self):
        return 1 - universo(self.eventoA, self.espaco_amostral).prob()

    def indepedent(self):
        # P(A ∩ B) = P(A) * P(B), válida se A e B forem independentes
        return universo(self.eventoA, self.espaco_amostral).prob() * universo(self.eventoB, self.espaco_amostral).prob()

    def conditional(self):
        # P(A|B) = P(A ∩ B) / P(B)
        prob_ab = universo(self.intercet(), self.espaco_amostral).prob()
        prob_b = universo(self.eventoB, self.espaco_amostral).prob()
        return prob_ab / prob_b

    def conditional_mult(self):
        # P(A ∩ B ∩ C) = P(A) * P(B|A) * P(C|A ∩ B)
        a = set(self.eventoA)
        ab = a & set(self.eventoB)
        abc = ab & set(self.eventoC)
        prob_a = len(a) / len(self.espaco_amostral)
        prob_b_dado_a = len(ab) / len(a)
        prob_c_dado_ab = len(abc) / len(ab)
        return prob_a * prob_b_dado_a * prob_c_dado_ab
```

### Funções auxiliares

```python
#números ímpares
def impar(lista):
   return list(filter(lambda x: (x%2 != 0) , lista))


#números pares
def par(lista):
   return list(filter(lambda x: (x%2 == 0) , lista))


#múltiplos de 5
def mult_cinco(lista):
   return list(filter(lambda x: (x%5 == 0) , lista))


#não múltiplos de 5
def nao_mult_cinco(lista):
   return list(filter(lambda x: (x%5 != 0) , lista))
```
