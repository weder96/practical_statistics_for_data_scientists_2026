## **Concepts**

### **Random Experiment**
A phenomenon whose outcome cannot be predicted: if the experiment is repeated under the same conditions, the result may be different.

### **Sample Space**
The sample space $S$ is the set of all possible outcomes of the experiment.

```python
# Sample space with 60 possible outcomes (0 to 59)
S = list(range(60))
len(S)
```

### **Events**
An event is a subset of the sample space—that is, a set of possible outcomes of the experiment.

### **Operations with Events**

* `Union:` the event $A \cup B$ contains all outcomes that belong to $A$, to $B$, or to both.

* `Intersection:` the event $A \cap B$ contains the outcomes that belong to both $A$ and $B$ at the same time.

* `Complement:` the event $A^c$ contains all outcomes in the sample space that do not belong to $A$.

### **Independent Events**
The occurrence of event $B$ does not change the probability of event $A$ occurring.

### **Mutually Exclusive Events**
Events $A$ and $B$ have no elements in common—that is, they cannot occur simultaneously ($A \cap B = \emptyset$).

## **Definition of Probability**
Probability measures the likelihood of an event $A$ occurring within the sample space. When all outcomes are equally likely:

$P(A) = \frac{n_{A}}{n}$

where $n_A$ is the number of outcomes favorable to $A$ and $n$ is the total number of possible outcomes.

```python
# Rolling a die: probability of getting an even number (in %)
die = [1, 2, 3, 4, 5, 6]
A = [2, 4, 6]

probability = len(A) / len(die) * 100
probability
```

## **Basic Rules**

1. `Probability range:` probability is always between 0 and 1 (0% to 100%), $0 \leq P(A) \leq 1$
2. `Probability of the sample space:` $P(S) = 1$
3. `Probability of the empty set:` $P(\emptyset) = 0$
4. `Addition rule:` $P(A \cup B) = P(A) + P(B) - P(A \cap B)$. If $A$ and $B$ are mutually exclusive, $P(A \cup B) = P(A) + P(B)$
5. `Complementary events:` $P(A^c) = 1 - P(A)$
6. `Multiplication rule for independent events:` $P(A \cap B) = P(A)\cdot P(B)$

## **Conditional Probability**
When events are not independent, we use conditional probability: given that $B$ has occurred, what is the probability that $A$ occurs?

$P(A|B) = \frac{P(A \cap B)}{P(B)}$

* `Multiplication rule:` $P(A \cap B) = P(A)\cdot P(B|A)$, which can be extended to three events: $P(A \cap B \cap C) = P(A)\cdot P(B|A)\cdot P(C|A \cap B)$

## **Bayes' Theorem**
Allows recalculating (updating) the probability of an event when new information becomes available. Let $B_1, ..., B_n$ be mutually exclusive events that partition the sample space:

$P(B_{i} | A) = \frac{P(A|B_i)\cdot P(B_i)}{\sum_{j=1}^{n}P(A|B_j)\cdot P(B_j)}$

## **Combinatorics**

Tools for counting the number of possible outcomes when the sample space is large.

* `Arrangement (Permutation of k from n):` number of ways to choose $k$ elements from $n$ when order matters. $A_{n,k} = \frac{n!}{(n-k)!}$
* `Combination:` number of ways to choose $k$ elements from $n$ when order does not matter. $C_{n,k} = \frac{n!}{k!(n-k)!}$
* `Permutation:` number of ways to arrange $n$ elements. $P_n = n!$

```python
# Example: 10 candidates for 3 positions
n, k = 10, 3

print(f"Arrangement (order matters): {math.perm(n, k)}")
print(f"Combination (order does not matter): {math.comb(n, k)}")
print(f"Permutation of 3 elements: {math.factorial(3)}")
```

A small class to calculate the probabilities covered above from lists of events.

```python
class Universe:
    def __init__(self, eventA, sample_space, eventB=None, eventC=None):
        self.eventA = eventA
        self.eventB = eventB
        self.eventC = eventC
        self.sample_space = sample_space

    def prob(self):
        return len(self.eventA) / len(self.sample_space)

    def intersect(self):
        return list(set(self.eventA).intersection(self.eventB))

    def union(self):
        return list(set(self.eventA).union(self.eventB))

    def addition_general(self):
        # P(A ∪ B) = P(A) + P(B) - P(A ∩ B)
        prob_a = Universe(self.eventA, self.sample_space).prob()
        prob_b = Universe(self.eventB, self.sample_space).prob()
        prob_ab = Universe(self.intersect(), self.sample_space).prob()
        return prob_a + prob_b - prob_ab

    def addition_mutually_exclusive(self):
        # P(A ∪ B) = P(A) + P(B), valid only if A and B are mutually exclusive
        prob_a = Universe(self.eventA, self.sample_space).prob()
        prob_b = Universe(self.eventB, self.sample_space).prob()
        return prob_a + prob_b

    def complement(self):
        return 1 - Universe(self.eventA, self.sample_space).prob()

    def independent(self):
        # P(A ∩ B) = P(A) * P(B), valid if A and B are independent
        return Universe(self.eventA, self.sample_space).prob() * Universe(self.eventB, self.sample_space).prob()

    def conditional(self):
        # P(A|B) = P(A ∩ B) / P(B)
        prob_ab = Universe(self.intersect(), self.sample_space).prob()
        prob_b = Universe(self.eventB, self.sample_space).prob()
        return prob_ab / prob_b

    def conditional_mult(self):
        # P(A ∩ B ∩ C) = P(A) * P(B|A) * P(C|A ∩ B)
        a = set(self.eventA)
        ab = a & set(self.eventB)
        abc = ab & set(self.eventC)
        prob_a = len(a) / len(self.sample_space)
        prob_b_given_a = len(ab) / len(a)
        prob_c_given_ab = len(abc) / len(ab)
        return prob_a * prob_b_given_a * prob_c_given_ab
```

### Helper Functions

```python
# Odd numbers
def odd(lst):
    return list(filter(lambda x: (x % 2 != 0), lst))


# Even numbers
def even(lst):
    return list(filter(lambda x: (x % 2 == 0), lst))


# Multiples of 5
def mult_five(lst):
    return list(filter(lambda x: (x % 5 == 0), lst))


# Non-multiples of 5
def not_mult_five(lst):
    return list(filter(lambda x: (x % 5 != 0), lst))
```
